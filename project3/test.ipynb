```python
import pandas as pd

data = {
    'Name':['Pavan', 'Nikil', 'Rohan', 'Rahul'],
    'Age':[23, None, 32, None],
    'Salary':[3000.0, 5600.0, 12000.0, 12345.0]
}

df = pd.DataFrame(data)
print("Original Data")
print(df)
```

```python
print(df.isnull().sum())
df_drop = df.dropna()
print(df_drop)
```

```python
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].mean(), inplace=True)
print(df)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
X = [[10], [20], [30], [40], [100000]]
y = [100, 150, 225, 356, 356]
model.fit(X, y)
hour  = float(input("Enter the hour: "))
predict = model.predict([[hour]])
print(f"Predicted value for {hour} hour is {predict[0]}")
```

```python
# Alternative approach: Test with multiple values without input()
from sklearn.linear_model import LinearRegression
model = LinearRegression()
X = [[10], [20], [30], [40]]
y = [100, 150, 225, 356]
model.fit(X, y)

# Test with multiple hour values
test_hours = [15, 25, 35, 45, 50]
for hour in test_hours:
    predict = model.predict([[hour]])
    print(f"Predicted value for {hour} hour is {predict[0]:.2f}")

# If you want to test with a specific hour, just change this value:
specific_hour = 42
predict = model.predict([[specific_hour]])
print(f"\nPredicted value for {specific_hour} hour is {predict[0]:.2f}")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

# Use the fixed CSV file
data = pd.read_csv('score_fixed.csv')
print("Data loaded successfully!")
print("Columns:", data.columns.tolist())
print("Data shape:", data.shape)
print("\nFirst few rows:")
print(data.head())

X = data[['Hours']]
y = data['Score']
model = LinearRegression()
model.fit(X, y)
predict_score = model.predict(X)

mse = mean_squared_error(y, predict_score)
mae = mean_absolute_error(y, predict_score)
rmse = np.sqrt(mse)
print(f"\nModel Performance:")
print(f"Mean Squared Error: {mse:.2f}")
print(f"Mean Absolute Error: {mae:.2f}")   
print(f"Root Mean Squared Error: {rmse:.2f}")

# Instead of using input(), let's test with a specific value
test_hours = 7.5
new_pred = model.predict([[test_hours]])
print(f"\nPredicted score for {test_hours} hours is {new_pred[0]:.2f}")

# You can change this value to test different hours:
# test_hours = 15  # Change this to any value you want to test
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

df = pd.read_csv("score1.csv")   
print("Data loaded successfully!")
print(df.head())  


X = df[['Hours']]
y = df['Score']

model = LinearRegression()
model.fit(X, y)

predict_score = model.predict(X)

mae = mean_absolute_error(y, predict_score)
mse = mean_squared_error(y, predict_score)
rmse = np.sqrt(mse)

print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"Root Mean Squared Error: {rmse}")

hour = float(input("Enter the hour: "))
new_pred = model.predict([[hour]])
print(f"Predicted Score for {hour} hours: {new_pred[0]}")
```
